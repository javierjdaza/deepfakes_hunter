```python
# Links
# https://drive.google.com/file/d/1QRxxjBIljQg3OtDM6TG7ONsTtjcKTOAT/view?usp=drive_link
```

```python
def download_from_gdrive(file_id, output_path):
    
    url = f"https://drive.google.com/uc?id={file_id}"
    gdown.download(url, output = output_path, quiet = False)
```

```python
file_id_1 = '1QRxxjBIljQg3OtDM6TG7ONsTtjcKTOAT'

for file_id in [file_id_1]:
    
    download_from_gdrive(file_id, './')
```
