# 01 · Face Swap Generation — ArcFace Embeddings

Extract 512-dim L2-normalised ArcFace embeddings for every aligned face. The resulting matrix `S = E Eᵀ` is used for pair selection in the next step.

```python
from PIL import Image
from io import BytesIO
from deepface import DeepFace
import numpy as np
import glob
import os
from tqdm import tqdm
```

```python
def extract_embeddings(img_array):
        
    embedding_objs = DeepFace.represent(
        img_path = img_array,
        model_name = 'VGG-Face',
        # model_name = 'ArcFace',
        enforce_detection = False,
        align = False
    )[0]["embedding"]

    return embedding_objs

def l2_normalize(x):
    x = np.array(x, dtype=np.float32)
    return x / np.linalg.norm(x)
```

#### FLICKR

```python
files = glob.glob('../00 ETL/FLICKR/cropped_and_aligned/*')
len(files)
```

```python
embeddings = []
image_names = []

for img_path in tqdm(files):

    img_name = os.path.basename(img_path)
    img_pillow = Image.open(img_path).convert("RGB")
    img_array = np.array(img_pillow)

    # --- Get Embeddings ---
    emb = extract_embeddings(img_array)
    emb = l2_normalize(emb)

    embeddings.append(emb)
    image_names.append(img_name)

embeddings = np.array(embeddings, dtype = np.float32)
image_names = np.array(image_names)

# embeddings.shape = (B, 512)
# image_names.shape = (B,)
```

#### Save

```python
np.savez('./face_embeddings_artifacts/flickr_face_embeddings_vgg.npz',
         embeddings = embeddings,
         image_names = image_names)
```
